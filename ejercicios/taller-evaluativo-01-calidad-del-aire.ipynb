## Ejercicio 2 — Conversión de unidades y error frente al equipo patrón

Convierte la lectura del sensor de mg/m³ a µg/m³ usando el factor_conversion del Ejercicio 1, y se calcula el error absoluto, el error relativo porcentual y el error cuadrático frente al valor del equipo patrón.

```python
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

lectura_sensor_ug = lectura_sensor_mg * factor_conversion
error_absoluto = lectura_sensor_ug - valor_patron_ug
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100
error_cuadratico = error_absoluto ** 2

print("Lectura convertida (ug/m3):", lectura_sensor_ug)
print("Error absoluto:", error_absoluto)
print("Error relativo porcentual:", error_relativo_porcentual)
print("Error cuadratico:", error_cuadratico)
```

## Ejercicio 3 — Cobertura temporal del lote descargado

Con numero_lecturas del Ejercicio 1 y 12 lecturas por hora, se calcula cuántas horas y días completos cubre el lote, usando solamente división entera y módulo.

```python
lecturas_por_hora = 12

horas_completas = numero_lecturas // lecturas_por_hora
lecturas_sobrantes = numero_lecturas % lecturas_por_hora

dias_completos = horas_completas // 24
horas_sobrantes = horas_completas % 24

print("Horas completas:", horas_completas)
print("Lecturas sobrantes:", lecturas_sobrantes)
print("Dias completos:", dias_completos)
print("Horas sobrantes:", horas_sobrantes)
```

## Ejercicio 4 — Validar una lectura sin decidir todavía

Evaluar si la lectura es el centinela de dato faltante, si cae dentro del rango físico del sensor, y si es una lectura válida (en rango y no faltante), combinando and y not sin usar if.

```python
lectura = 41.8
codigo_dato_faltante = -999.0
limite_superior_sensor = 500.0

es_dato_faltante = lectura == codigo_dato_faltante
esta_en_rango = lectura >= 0 and lectura <= limite_superior_sensor
lectura_valida = esta_en_rango and not es_dato_faltante

print("Es dato faltante:", es_dato_faltante)
print("Esta en rango:", esta_en_rango)
print("Lectura valida:", lectura_valida)
```

## Ejercicio 5 — Clasificar la lectura en la escala de calidad del aire

lectura del Ejercicio 4 se clasifica la concentración según la escala oficial de PM2.5, usando una única cadena if / elif / else ordenada de menor a mayor umbral.

```python
if lectura <= 12.0:
    categoria = "Buena"
elif lectura <= 35.4:
    categoria = "Moderada"
elif lectura <= 55.4:
    categoria = "Dañina para grupos sensibles"
else:
    categoria = "Dañina para la salud"

print("Categoria:", categoria)
```

## Ejercicio 6 — Limpiar la serie horaria

Se recorre la serie de doce lecturas descartando los centinelas de dato faltante (-999.0) con continue, acumulando las lecturas válidas en una lista nueva y calculando el promedio solo sobre esas lecturas.

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]

lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for valor in lecturas_pm25:
    if valor == -999.0:
        lecturas_descartadas += 1
        continue
    lecturas_validas.append(valor)
    suma_pm25 += valor

promedio_pm25 = suma_pm25 / len(lecturas_validas)

print("Lecturas validas:", len(lecturas_validas))
print("Lecturas descartadas:", lecturas_descartadas)
print("Promedio PM2.5:", promedio_pm25)
```

## Ejercicio 7 — Estadística descriptiva de la serie limpia

Con dos recorridos sobre lecturas_validas (uno para los extremos, otro para la dispersión que depende del promedio) se calculan el máximo, el mínimo, el rango y la desviación estándar, sin usar max(), min() ni sum().

```python
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for valor in lecturas_validas:
    if valor > maximo_pm25:
        maximo_pm25 = valor
    if valor < minimo_pm25:
        minimo_pm25 = valor

rango_pm25 = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0
for valor in lecturas_validas:
    suma_desviaciones += (valor - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print("Maximo:", maximo_pm25)
print("Minimo:", minimo_pm25)
print("Rango:", rango_pm25)
print("Desviacion estandar:", desviacion_estandar)
```

## Ejercicio 8 — Decaimiento tras cerrar la fuente de emisión

Tras alcanzar el pico de contaminación, la concentración baja un 15 % cada hora. Con while se calcula cuántas horas transcurren hasta que la concentración vuelve a categoría "Buena" (≤ 12.0 µg/m³).

```python
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0

horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion = concentracion * factor_decaimiento
    horas_transcurridas += 1

print("Horas transcurridas:", horas_transcurridas)
print("Concentracion final:", concentracion)
```

## Ejercicio 9 — Registros de varias estaciones

Se recorren los registros de las tres estaciones (diccionarios con tupla de coordenadas y lista de lecturas propia), desempaquetando las coordenadas, calculando el promedio de cada estación con un bucle interno, obteniendo la humedad con get (para no fallar en las estaciones sin ese sensor) y clasificando cada promedio con la misma escala del Ejercicio 5.

```python
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte",  "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur",    "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

for estacion in registros_estaciones:
    latitud_estacion, longitud_estacion = estacion["coordenadas"]

    suma_lecturas_estacion = 0.0
    for lectura_estacion in estacion["lecturas"]:
        suma_lecturas_estacion += lectura_estacion
    promedio_estacion = suma_lecturas_estacion / len(estacion["lecturas"])

    humedad_estacion = estacion.get("humedad_relativa", "no disponible")

    if promedio_estacion <= 12.0:
        categoria_estacion = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_estacion = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_estacion = "Dañina para grupos sensibles"
    else:
        categoria_estacion = "Dañina para la salud"

    print(estacion["codigo"], "-", estacion["sector"],
          "- lat:", latitud_estacion, "lon:", longitud_estacion,
          "- promedio:", promedio_estacion,
          "- categoria:", categoria_estacion,
          "- humedad:", humedad_estacion)
```
