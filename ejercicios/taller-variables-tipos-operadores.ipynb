## Ejercicio 1 — Inventario de la tienda

Se declaran siete variables que representan el inventario de una tienda pequeña: el nombre de un producto (`str`), el precio de tres productos (`int` para cuaderno y lápiz, `float` para café), si hay descuentos activos (`bool`) y las cantidades a comprar (`int`). Al final se verifica el tipo de cada variable con `type()`.

```python
# Ejercicio 1 — Inventario de la tienda

nombre_producto = "Cuaderno"
precio_cuaderno = 3500
precio_lapiz = 800
precio_cafe = 6200.50
hay_descuento = False
cantidad_cuadernos = 3
cantidad_lapices = 5

print(type(nombre_producto))
print(type(precio_cuaderno))
print(type(precio_lapiz))
print(type(precio_cafe))
print(type(hay_descuento))
print(type(cantidad_cuadernos))
print(type(cantidad_lapices))
```

## Ejercicio 2 — Total de la compra y descuento

Con las variables del Ejercicio 1 se calculan los subtotales por producto, el total general (que incluye el precio del café), un descuento del 10 % y el total final con descuento aplicado. Se usan los operadores `*`, `+` y `-`.

```python
# Ejercicio 2 — Total de la compra y descuento

total_cuadernos = precio_cuaderno * cantidad_cuadernos
print("Total cuadernos:", total_cuadernos)

total_lapices = precio_lapiz * cantidad_lapices
print("Total lápices:", total_lapices)

total_general = total_cuadernos + total_lapices + precio_cafe
descuento = total_general * 0.10
total_con_descuento = total_general - descuento

print("Total general:", total_general)
print("Descuento (10%):", descuento)
print("Total con descuento:", total_con_descuento)
```

## Ejercicio 3 — Reparto en paquetes

Con `cantidad_cuadernos` se calcula cuántos paquetes completos de 4 unidades se pueden armar (división entera `//`) y cuántos cuadernos sobran (módulo `%`). También se eleva el precio del café al cuadrado usando el operador de potencia `**`.

```python
# Ejercicio 3 — Reparto en paquetes

paquetes_completos = cantidad_cuadernos // 4
cuadernos_sueltos = cantidad_cuadernos % 4
precio_cafe_cuadrado = precio_cafe ** 2

print("Paquetes completos de 4 cuadernos:", paquetes_completos)
print("Cuadernos sueltos:", cuadernos_sueltos)
print("Precio del café al cuadrado:", precio_cafe_cuadrado)
```

## Ejercicio 4 — Comparar precios de café

Se declaran los precios de dos marcas de café y se comparan usando operadores de comparación: `<` para determinar cuál es más barata y `==` para verificar si cuestan lo mismo.

```python
# Ejercicio 4 — Comparar precios de café

precio_marca_a = 15900
precio_marca_b = 14500

print("¿Marca A es más barata que Marca B?", precio_marca_a < precio_marca_b)
print("¿Ambas marcas cuestan lo mismo?", precio_marca_a == precio_marca_b)
```

## Ejercicio 5 — Promedio y comparación de temperaturas

Se declaran las temperaturas de tres días de la semana y una temperatura de referencia (todas `float`). Se calcula el promedio de los tres días con `/`, la diferencia entre martes y lunes, y se realizan tres comparaciones: si el martes superó el promedio de referencia, si el miércoles fue igual y si el lunes fue el día más frío.

```python
# Ejercicio 5 — Promedio y comparación de temperaturas

temp_lunes = 18.5
temp_martes = 21.0
temp_miercoles = 19.8
temp_promedio_semana = 20.0

promedio_tres_dias = (temp_lunes + temp_martes + temp_miercoles) / 3
print("Promedio de los tres días:", promedio_tres_dias)

diferencia_martes_lunes = temp_martes - temp_lunes
print("Diferencia martes - lunes:", diferencia_martes_lunes)

print("¿El martes fue más caluroso que el promedio de referencia?", temp_martes > temp_promedio_semana)
print("¿El miércoles fue igual al promedio de referencia?", temp_miercoles == temp_promedio_semana)
print("¿El lunes fue el día más frío?", temp_lunes < temp_martes and temp_lunes < temp_miercoles)
```

## Ejercicio 6 — Temperatura templada y día caluroso

Con las variables de temperatura del Ejercicio 5 se evalúan dos condiciones compuestas: si la temperatura del lunes es "templada" (entre 15 y 25 grados, usando `and`) y si al menos uno de los tres días superó los 20 grados (usando `or`).

```python
# Ejercicio 6 — Temperatura templada y día caluroso

lunes_templado = temp_lunes >= 15 and temp_lunes <= 25
print("¿El lunes tuvo temperatura templada (15–25°C)?", lunes_templado)

hubo_dia_caluroso = temp_lunes > 20 or temp_martes > 20 or temp_miercoles > 20
print("¿Al menos un día superó los 20°C?", hubo_dia_caluroso)
```

## Ejercicio 7 — Factura de servicios y consumo normal

Se declaran el consumo en kWh (`int`) y la tarifa por kWh (`float`). Se calcula el total a pagar con `*` y se evalúa si el consumo está en el rango considerado "normal" por el proveedor (entre 100 y 200 kWh), combinando dos comparaciones con `and`.

```python
# Ejercicio 7 — Factura de servicios y consumo normal

consumo_kwh = 142
tarifa_kwh = 623.5

total_factura = consumo_kwh * tarifa_kwh
print("Total a pagar:", total_factura)

consumo_normal = consumo_kwh >= 100 and consumo_kwh <= 200
print("¿El consumo es normal (100–200 kWh)?", consumo_normal)
```

## Ejercicio 8 — Promedio de calificaciones y aprobación

```python
# Ejercicio 8 — Promedio de calificaciones y aprobación

nota_parcial_1 = 4.2
nota_parcial_2 = 3.5
nota_parcial_3 = 4.8

promedio_notas = (nota_parcial_1 + nota_parcial_2 + nota_parcial_3) / 3
print("Promedio de calificaciones:", promedio_notas)

print("¿El estudiante aprobó (promedio >= 3.0)?", promedio_notas >= 3.0)
```
