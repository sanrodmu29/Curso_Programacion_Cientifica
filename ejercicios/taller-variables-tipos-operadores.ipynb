## Taller — Variables, Tipos y Operadores en Python
Este notebook contiene diez ejercicios de dificultad creciente sobre declaración de variables, tipos de datos (int, float, str, bool), operadores aritméticos, de comparación y lógicos.

### Ejercicio 1 — Inventario de la tienda
Se declaran siete variables que representan el inventario de una tienda: nombre de producto (str), precios (int y float), si hay descuentos (bool) y cantidades (int). Se verifica el tipo de cada una con `type()`.

```python
nombre_producto = "Cuaderno"
precio_cuaderno = 3500
precio_lapiz = 800
precio_cafe = 6200.50
hay_descuento = False
cantidad_cuadernos = 3
cantidad_lapices = 5

print(type(nombre_producto))
print(type(precio_cuaderno))
print(type(precio_lapiz))
print(type(precio_cafe))
print(type(hay_descuento))
print(type(cantidad_cuadernos))
print(type(cantidad_lapices))
```

### Ejercicio 2 — Total de la compra y descuento
Con las variables del Ejercicio 1 se calculan subtotales, el total general, un descuento del 10 % y el total con descuento. Se usan los operadores `*`, `+` y `-`.

```python
total_cuadernos = precio_cuaderno * cantidad_cuadernos
print("Total cuadernos:", total_cuadernos)

total_lapices = precio_lapiz * cantidad_lapices
print("Total lápices:", total_lapices)

total_general = total_cuadernos + total_lapices + precio_cafe
descuento = total_general * 0.10
total_con_descuento = total_general - descuento

print("Total general:", total_general)
print("Descuento (10%):", descuento)
print("Total con descuento:", total_con_descuento)
```

### Ejercicio 3 — Reparto en paquetes
Con `cantidad_cuadernos` se calcula cuántos paquetes completos de 4 unidades se pueden armar (`//`), cuántos sobran (`%`) y el precio del café al cuadrado (`**`).

```python
paquetes_completos = cantidad_cuadernos // 4
cuadernos_sueltos = cantidad_cuadernos % 4
precio_cafe_cuadrado = precio_cafe ** 2

print("Paquetes completos:", paquetes_completos)
print("Cuadernos sueltos:", cuadernos_sueltos)
print("Precio café al cuadrado:", precio_cafe_cuadrado)
```

### Ejercicio 4 — Comparar precios de café
Se declaran los precios de dos marcas de café y se comparan con `<` y `==` para saber cuál es más barata y si cuestan lo mismo.

```python
precio_marca_a = 15900
precio_marca_b = 14500

print("¿Marca A más barata que Marca B?", precio_marca_a < precio_marca_b)
print("¿Ambas cuestan lo mismo?", precio_marca_a == precio_marca_b)
```

### Ejercicio 5 — Promedio y comparación de temperaturas
Se declaran temperaturas de tres días y una referencia (float). Se calcula el promedio con `/`, la diferencia martes-lunes, y tres comparaciones.

```python
temp_lunes = 18.5
temp_martes = 21.0
temp_miercoles = 19.8
temp_promedio_semana = 20.0

promedio_tres_dias = (temp_lunes + temp_martes + temp_miercoles) / 3
print("Promedio tres días:", promedio_tres_dias)

diferencia_martes_lunes = temp_martes - temp_lunes
print("Diferencia martes - lunes:", diferencia_martes_lunes)

print("¿Martes más caluroso que referencia?", temp_martes > temp_promedio_semana)
print("¿Miércoles igual a referencia?", temp_miercoles == temp_promedio_semana)
print("¿Lunes fue el más frío?", temp_lunes < temp_martes and temp_lunes < temp_miercoles)
```

### Ejercicio 6 — Temperatura templada y día caluroso
Con las temperaturas del Ejercicio 5 se evalúa si el lunes fue templado y si al menos un día superó los 20°C.

```python
lunes_templado = temp_lunes >= 15 and temp_lunes <= 25
print("¿Lunes con temperatura templada (15–25°C)?", lunes_templado)

hubo_dia_caluroso = temp_lunes > 20 or temp_martes > 20 or temp_miercoles > 20
print("¿Al menos un día superó los 20°C?", hubo_dia_caluroso)
```

### Ejercicio 7 — Factura de servicios y consumo normal
Se calcula el total a pagar por consumo eléctrico y se evalúa si el consumo está en el rango normal de 100 a 200 kWh.

```python
consumo_kwh = 142
tarifa_kwh = 623.5

total_factura = consumo_kwh * tarifa_kwh
print("Total a pagar:", total_factura)

consumo_normal = consumo_kwh >= 100 and consumo_kwh <= 200
print("¿Consumo normal (100–200 kWh)?", consumo_normal)
```

### Ejercicio 8 — Promedio de calificaciones y aprobación
Se calculan tres notas parciales, su promedio y se evalúa si el estudiante aprobó (promedio >= 3.0).

```python
nota_parcial_1 = 4.2
nota_parcial_2 = 3.5
nota_parcial_3 = 4.8

promedio_notas = (nota_parcial_1 + nota_parcial_2 + nota_parcial_3) / 3
print("Promedio de calificaciones:", promedio_notas)
print("¿Aprobó (>= 3.0)?", promedio_notas >= 3.0)
```

### Ejercicio 9 — Aprobación y asistencia
Se retoma el promedio del Ejercicio 8 y se usa el operador `not` para mostrar si el estudiante NO reprobó.

```python
asistencia_porcentaje = 85.0

reprobado = promedio_notas < 3.0
print("¿El estudiante NO reprobó?", not reprobado)
```

### Ejercicio 10 — Intercambiar dos precios
Se intercambian los valores de dos variables usando una variable temporal.

```python
precio_tornillo = 250
precio_tuerca = 180

print("Antes del intercambio:")
print("precio_tornillo:", precio_tornillo)
print("precio_tuerca:", precio_tuerca)

temporal = precio_tornillo
precio_tornillo = precio_tuerca
precio_tuerca = temporal

print("Después del intercambio:")
print("precio_tornillo:", precio_tornillo)
print("precio_tuerca:", precio_tuerca)
```
